```python
from langchain_core.messages import AIMessage, HumanMessage
from pprint import pprint
from langchain_groq import ChatGroq
```

```python
messages = []
messages.append(HumanMessage(content="Hello, I am Kunal", name= "Kunal"))
messages.append(AIMessage(content="Hi Kunal, How May I Help You", name= "AI Agent"))
messages.append(HumanMessage(content="I need to Learn Programing Language", name = "Kunal"))

for item in messages:
    item.pretty_print()
```

```python
llm = ChatGroq(model = "qwen/qwen3-32b")
llm.invoke(messages).response_metadata
```
